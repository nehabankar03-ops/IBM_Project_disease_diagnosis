# Diagnosis Prediction Project
**Problem Statement:** To predict diagnosis from patient-related clinical features using machine learning.

This notebook loads a dataset, performs preprocessing, trains classification models, evaluates them, and demonstrates diagnosis prediction for a new patient record.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Change this path to your dataset file
DATA_PATH = 'diagnosis_dataset.csv'
df = pd.read_csv(DATA_PATH)
print(df.head())
print(df.shape)
print(df.info())
```

```python
# Set the target column name used by your dataset
TARGET = 'Diagnosis'

X = df.drop(columns=[TARGET])
y = df[TARGET]

numeric_cols = X.select_dtypes(include=['number']).columns.tolist()
categorical_cols = X.select_dtypes(exclude=['number']).columns.tolist()

numeric_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
categorical_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer([
    ('num', numeric_pipe, numeric_cols),
    ('cat', categorical_pipe, categorical_cols)
])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(n_estimators=200, random_state=42))
])
model.fit(X_train, y_train)
print('Model trained successfully.')
```

```python
y_pred = model.predict(X_test)
print('Accuracy:', accuracy_score(y_test, y_pred))
print('\nClassification Report:\n', classification_report(y_test, y_pred))
print('\nConfusion Matrix:\n', confusion_matrix(y_test, y_pred))
```

```python
# Example prediction using one test record
sample = X_test.iloc[[0]]
prediction = model.predict(sample)[0]
print('Predicted diagnosis:', prediction)
```

## Conclusion
The Random Forest classification pipeline predicts the diagnosis class from the available patient features. The final accuracy and classification report depend on the dataset used. This project is intended for academic/educational use and is not a medical diagnosis tool.
